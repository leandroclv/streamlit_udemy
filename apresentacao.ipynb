###Steramlit

- Criar aplicações de dados web
- Não requer conhecimento de programação web
- Cada nova requisição toda a aplicação é executada
- Cache: permite salvar em memória objetos
- Session: permite compartilhar objetos ba aplicação, mesmo em diferentes páginas

####Como funciona

- Você importa a biblioteca streamlit
- Roda a aplicação na linha decomando:
    - python -m streamlit run app.py
- Pronto!
- Você pode publicar a sua aplicação:
    - Community cloud
    - Craiar seu próprio servidor

####Comandos

- st.write(): renderiza objetos: tabelas, textos, etc.
- Comandos mágicos: redenriza objetos automaticamente, mesmo sem usar st.write()
- Comandos específicos: mais parâmetros, renderiza de outra forma, podem er modificados
    - st.dataframe(): dataframe com várias opções
    - st.table(): tabela estática

####Criando o ambiente virtual (terminal)
- python -m venv <nome_do_ambiente>

####Ativar o ambiente virtual
- .\<nome_do_ambiente>\Scripts\activate

####Instalar as bibliotecas no ambiente virtual
- pip install -r <bibliotecas com o arquivo requirements.txt>
- ativa novamente (passo anterior)
