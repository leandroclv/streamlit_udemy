###Problema do Cliente

- Um revenda de carros quer poder prever a qualidade do carro a partir de dados de carros anteriores
- Assim cla pode precificar de forma mais adequada o serviço de assistência 24 horas
- A empresa possui uma grande base de dados de veículos adiquiridos anteriormente, com características como exigência de manutenção, portas, tamanho do porta malas, segurança, etc.

Cliente tem dados históricos

- Podemos usar Machine Learning, tarefa de classificação
- Machine Learning são técnicas de prever ou classificar algo baseados em dados históricos
